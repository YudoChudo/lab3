```python
import json

purchases = {}

with open('purchase_log.txt', encoding='utf-8') as file:
    for line in file:
        line = line.strip()
        if not line:
            continue
        data = json.loads(line)
        purchases[data['user_id']] = data['category']

for user_id, category in list(purchases.items())[:10]:
    print(user_id, category)
```

```python
import csv
import json
from itertools import islice

purchases = {}

with open('purchase_log.txt', encoding='utf-8') as purchase_file:
    for line in purchase_file:
        line = line.strip()
        if not line:
            continue
        purchase = json.loads(line)
        purchases[purchase['user_id']] = purchase['category']

with open('visit_log.csv', encoding='utf-8') as visit_file, \
     open('funnel.csv', 'w', encoding='utf-8', newline='') as funnel_file:

    reader = csv.reader(visit_file)
    writer = csv.writer(funnel_file)

    writer.writerow(['user_id', 'source', 'category'])

    first_row = next(reader, None)

    if first_row:
        if first_row[0] == 'user_id' and len(first_row) > 1 and first_row[1] == 'source':
            pass
        else:
            user_id = first_row[0]
            source = first_row[1]
            if user_id in purchases:
                writer.writerow([user_id, source, purchases[user_id]])

    for row in reader:
        if not row:
            continue
        user_id = row[0]
        source = row[1]
        if user_id in purchases:
            writer.writerow([user_id, source, purchases[user_id]])

with open('funnel.csv', encoding='utf-8') as file:
    for line in islice(file, 10):
        print(line.rstrip())
```
